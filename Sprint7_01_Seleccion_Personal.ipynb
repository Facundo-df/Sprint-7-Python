Sprint 7.01 - Ejercicio 1: Selección de Personal

Objetivo del ejercicio

Desarrollar un sistema de inscripción y selección de personal para el departamento de Recursos Humanos, permitiendo registrar un máximo de diez candidatos válidos.

El programa debe solicitar y validar los datos de cada candidato, calcular su puntuación según los criterios establecidos, asignarle una clasificación y generar un informe final ordenado por prioridad y puntuación.

Para resolverlo, divido el programa en funciones, cada una con una responsabilidad específica.

## 1. Función `obtenir_dades()`

1. Solicito el nombre y apellido del candidato mediante `input()` y lo almaceno en `nombre`.
2. Solicito el teléfono y lo almaceno en `telefono`.
3. Solicito los años de experiencia y los almaceno en `experiencia`.
4. Solicito el nivel de catalán y lo almaceno en `catalan`.
5. Solicito el nivel de Excel y lo almaceno en `excel`.
6. Solicito la titulación académica y la almaceno en `titulacion`.
7. Devuelvo los seis valores mediante `return` para que puedan ser validados.

```python
def obtenir_dades():
    nombre = input("Nombre y apellidos: ")
    telefono = input("Teléfono de contacto: ")
    experiencia = input("Años de experiencia: ")
    catalan = input("Nivel de catalán (0-3): ")
    excel = input("Nivel de Excel (0-3): ")
    titulacion = input("Titulación académica (0-3): ")

    return nombre, telefono, experiencia, catalan, excel, titulacion
```

```python
obtenir_dades()
```

## 2. Función `validar_dades()`

1. Recibo los seis datos introducidos por el candidato.
2. Elimino los espacios sobrantes del nombre y teléfono utilizando `strip()`.
3. Compruebo que el nombre no esté vacío.
4. Compruebo que el teléfono contenga exactamente nueve dígitos mediante `len()` e `isdigit()`.
5. Intento convertir la experiencia, el catalán, Excel y la titulación a números enteros mediante `int()`.
6. Si ocurre un error de conversión, muestro un mensaje y devuelvo `None`.
7. Compruebo que los años de experiencia no sean negativos.
8. Compruebo que los niveles de catalán, Excel y titulación estén entre 0 y 3.
9. Si algún valor no cumple las condiciones, muestro el error y devuelvo `None`.
10. Si todos los datos son correctos, devuelvo los seis valores validados.

```python
def validar_dades(nombre, telefono, experiencia, catalan, excel, titulacion):

    # 1. Limpiar los textos
    nombre = nombre.strip()
    telefono = telefono.strip()

    # 2. Validar el nombre
    if nombre == "":
        print("Error: el nombre no puede estar vacío.")
        return None

    # 3. Validar el teléfono
    if len(telefono) != 9 or not telefono.isdigit():
        print("Error: el teléfono debe tener 9 dígitos.")
        return None

    # 4. Convertir los datos numéricos
    try:
        experiencia = int(experiencia)
        catalan = int(catalan)
        excel = int(excel)
        titulacion = int(titulacion)

    except (ValueError, TypeError):
        print("Error: introduce valores numéricos válidos.")
        return None

    # 5. Validar la experiencia
    if experiencia < 0:
        print("Error: la experiencia no puede ser negativa.")
        return None

    # 6. Validar los niveles
    if not (0 <= catalan <= 3):
        print("Error: el nivel de catalán debe estar entre 0 y 3.")
        return None

    if not (0 <= excel <= 3):
        print("Error: el nivel de Excel debe estar entre 0 y 3.")
        return None

    if not (0 <= titulacion <= 3):
        print("Error: la titulación debe estar entre 0 y 3.")
        return None

    # 7. Devolver los datos correctos
    return nombre, telefono, experiencia, catalan, excel, titulacion
```

```python
validar_dades("Laura Serra", "608112233", "6", "3", "3", "3")
```

```python
validar_dades("Candidat Error", "611000000", 5, 7, 3, 2)
```

## 3. Función `calcular_puntuacio()`

1. Recibo los años de experiencia y las puntuaciones de catalán, Excel y titulación.
2. Multiplico los años de experiencia por 1.5.
3. Sumo al resultado los puntos de catalán, Excel y titulación.
4. Almaceno el total en la variable `puntuacion`.
5. Devuelvo la puntuación calculada mediante `return`.

```python
def calcular_puntuacio(experiencia, catalan, excel, titulacion):
    puntuacion = (experiencia * 1.5) + catalan + excel + titulacion
    return puntuacion
```

```python
resultado = calcular_puntuacio(6, 3, 3, 3)
print(resultado)
```

## 4. Función `classificar_candidat()`

1. Recibo la puntuación total del candidato.
2. Si la puntuación es mayor o igual a 12, devuelvo `"Prioridad Alta"`.
3. Si no se cumple lo anterior, compruebo si es mayor o igual a 8 y devuelvo `"Candidato adecuado"`.
4. Si no se cumple ninguna de las condiciones anteriores, devuelvo `"No prioritario"`.

```python
def classificar_candidat(puntuacion):
    if puntuacion >= 12:
        return "Prioridad Alta"
    elif puntuacion >= 8:
        return "Candidato adecuado"
    else:
        return "No prioritario"
```

```python
print(classificar_candidat(18))
print(classificar_candidat(10.5))
print(classificar_candidat(6))
```

## 5. Función `mostrar_individual()`

1. Recibo el nombre, la puntuación y la clasificación del candidato.
2. Muestro un encabezado para identificar el resultado individual.
3. Muestro el nombre del candidato.
4. Muestro la puntuación obtenida.
5. Muestro la clasificación asignada.
6. Utilizo `print()` y f-strings para presentar la información de forma clara.

```python
def mostrar_individual(nombre, puntuacion, clasificacion):
    print("\n--- RESULTADO INDIVIDUAL ---")
    print(f"Nombre: {nombre}")
    print(f"Puntuación: {puntuacion} puntos")
    print(f"Clasificación: {clasificacion}")
```

```python
mostrar_individual("Laura Serra", 18.0, "Prioridad Alta")
```

## 6. Función `mostrar_informe()`

1. Recibo la lista de candidatos registrados.
2. Ordeno la lista utilizando `sorted()`, tomando la puntuación como criterio mediante `key`.
3. Utilizo `reverse=True` para colocar las puntuaciones más altas primero. Como las categorías dependen de la puntuación, este orden también respeta las prioridades.
4. Guardo el resultado en `candidatos_ordenados`.
5. Muestro el encabezado del informe final.
6. Recorro la lista ordenada mediante un bucle `for`.
7. En cada iteración, accedo a las claves del diccionario correspondiente al candidato.
8. Muestro el nombre, teléfono, puntuación y clasificación utilizando `print()` y f-strings.

```python
def mostrar_informe(candidatos):

    candidatos_ordenados = sorted(
        candidatos,
        key=lambda candidato: candidato["puntuacion"],
        reverse=True
    )

    print("\n========== INFORME FINAL ==========")

    for candidato in candidatos_ordenados:
        print(f"Nombre: {candidato['nombre']}")
        print(f"Teléfono: {candidato['telefono']}")
        print(f"Puntuación: {candidato['puntuacion']}")
        print(f"Clasificación: {candidato['clasificacion']}")
        print("-----------------------------------")
```

```python
candidatos_prueba = [
    {
        "nombre": "Sergio López",
        "telefono": "644112211",
        "puntuacion": 10.5,
        "clasificacion": "Candidato adecuado"
    },
    {
        "nombre": "David Ferrer",
        "telefono": "677665544",
        "puntuacion": 6.0,
        "clasificacion": "No prioritario"
    },
    {
        "nombre": "Laura Serra",
        "telefono": "608112233",
        "puntuacion": 18.0,
        "clasificacion": "Prioridad Alta"
    }
]

mostrar_informe(candidatos_prueba)
```

## 7. Algoritmo principal: `main_seleccio()`

1. Creo una lista vacía llamada `candidatos` para almacenar las inscripciones válidas.
2. Utilizo un bucle `while` para repetir el proceso mientras haya menos de diez candidatos registrados.
3. Solicito los datos mediante `obtenir_dades()`.
4. Compruebo los datos mediante `validar_dades()`.
5. Si los datos son incorrectos, utilizo `continue` para volver al inicio del bucle sin registrar al candidato.
6. Si los datos son correctos, calculo la puntuación mediante `calcular_puntuacio()`.
7. Obtengo la clasificación mediante `classificar_candidat()`.
8. Creo un diccionario con el nombre, teléfono, puntuación y clasificación.
9. Agrego el diccionario a la lista mediante `append()`.
10. Muestro los resultados del candidato mediante `mostrar_individual()`.
11. Cuando la lista alcanza diez candidatos válidos, finalizo el bucle y genero el informe mediante `mostrar_informe()`.

```python
def main_seleccio():

    # 1. Crear la lista de candidatos
    candidatos = []

    # 2. Repetir hasta registrar 10 candidatos válidos
    while len(candidatos) < 10:

        print(f"\nInscripción {len(candidatos) + 1} de 10")

        # 3. Solicitar los datos
        datos = obtenir_dades()

        # 4. Validar los datos
        datos_validados = validar_dades(*datos)

        # 5. Rechazar registros incorrectos
        if datos_validados is None:
            continue

        # 6. Separar los datos validados
        nombre, telefono, experiencia, catalan, excel, titulacion = datos_validados

        # 7. Calcular puntuación y clasificación
        puntuacion = calcular_puntuacio(
            experiencia, catalan, excel, titulacion
        )

        clasificacion = classificar_candidat(puntuacion)

        # 8. Crear el diccionario del candidato
        candidato = {
            "nombre": nombre,
            "telefono": telefono,
            "puntuacion": puntuacion,
            "clasificacion": clasificacion
        }

        # 9. Guardar el candidato
        candidatos.append(candidato)

        # 10. Mostrar su resultado individual
        mostrar_individual(nombre, puntuacion, clasificacion)

    # 11. Mostrar el informe final
    mostrar_informe(candidatos)
```
