Función: obtenir_segment_i_factor()
1. Recibir el código de comportamiento del cliente.
2. Crear un diccionario donde cada código tenga asociado un segmento, un mensaje publicitario y un factor de riesgo.
3. Buscar en el diccionario la información correspondiente al código recibido.
4. Obtener el segmento, el mensaje y el factor.
5. Devolver los tres valores para utilizarlos en las siguientes funciones.

```python
def obtenir_segment_i_factor(codi):

    comportaments = {
        0: {
            "segment": "Navegador",
            "missatge": "Descubre nuestros productos.",
            "factor": 0.10
        },
        1: {
            "segment": "Interesado",
            "missatge": "Tus productos te están esperando.",
            "factor": 0.25
        },
        2: {
            "segment": "Compra pendiente",
            "missatge": "Finaliza tu compra.",
            "factor": 0.50
        },
        3: {
            "segment": "Abandono reciente",
            "missatge": "Tu carrito sigue disponible.",
            "factor": 0.75
        },
        4: {
            "segment": "Abandono antiguo",
            "missatge": "Recupera los productos de tu carrito.",
            "factor": 0.90
        }
    }

    dades = comportaments[codi]

    return dades["segment"], dades["missatge"], dades["factor"]
```

```python
segment, missatge, factor = obtenir_segment_i_factor(4)

print("Segmento:", segment)
print("Mensaje:", missatge)
print("Factor:", factor)
```

Función: obtenir_factor_edat()
1. Recibir el grupo de edad del cliente.
2. Crear un diccionario que relacione cada grupo de edad con su factor correspondiente.
3. Buscar el grupo recibido dentro del diccionario.
4. Devolver el factor de edad correspondiente.

```python
def obtenir_factor_edat(grup_edat):

    factors_edat = {
        "18-25": 1.20,
        "26-40": 1.00,
        "41-60": 0.85,
        "60+": 0.70
    }

    return factors_edat[grup_edat]
```

```python
print(obtenir_factor_edat("18-25"))
print(obtenir_factor_edat("26-40"))
print(obtenir_factor_edat("41-60"))
print(obtenir_factor_edat("60+"))
```

Función: calcular_risc()
1. Recibir el factor de comportamiento y el factor de edad.
2. Multiplicar ambos factores para calcular el riesgo final.
3. Guardar el resultado en la variable risc_final.
4. Devolver el riesgo calculado.

```python
def calcular_risc(factor_codi, factor_edat):
    risc_final = factor_codi * factor_edat
    return risc_final
```

```python
print(calcular_risc(0.90, 1.20))
```

Función: processar_clients()
1. Recibir la lista de clientes.
2. Crear una lista vacía llamada resultats.
3. Recorrer cada cliente utilizando un bucle for.
4. Obtener el segmento, mensaje y factor de comportamiento llamando a obtenir_segment_i_factor().
5. Obtener el factor de edad llamando a obtenir_factor_edat().
6. Calcular el riesgo final llamando a calcular_risc().
7. Crear un nuevo diccionario con los datos originales y los resultados calculados.
8. Añadir el diccionario a la lista resultats mediante append().
9. Repetir el proceso hasta terminar todos los clientes.
10. Devolver la lista completa.

```python
def processar_clients(clients):

    resultats = []

    for client in clients:

        segment, missatge, factor_codi = obtenir_segment_i_factor(
            client["codi_comportament"]
        )

        factor_edat = obtenir_factor_edat(client["grup_edat"])

        risc_final = calcular_risc(factor_codi, factor_edat)

        client_processat = {
            "id_cliente": client["id_client"],
            "nombre": client["nom"],
            "edad": client["edat"],
            "grupo_edad": client["grup_edat"],
            "codigo_comportamiento": client["codi_comportament"],
            "segment": segment,
            "missatge": missatge,
            "factor_codi": factor_codi,
            "factor_edat": factor_edat,
            "risc_final": risc_final
        }

        resultats.append(client_processat)

    return resultats
```

### Función `mostrar_diccionari()`

1. Recibo la lista de clientes procesados.
2. Muestro un encabezado para identificar los resultados.
3. Recorro la lista mediante un bucle `for`.
4. En cada iteración, accedo a los datos del cliente utilizando las claves de su diccionario.
5. Muestro los datos originales, el segmento, el mensaje y los factores utilizados.
6. Muestro el riesgo final redondeado a tres decimales.
7. Separo los registros para facilitar su lectura.

```python
def mostrar_diccionari(resultats):

    print("\n========== RESULTADOS DE MARKETING ==========")

    for client in resultats:
        print(f"ID: {client['id_cliente']}")
        print(f"Nombre: {client['nombre']}")
        print(f"Edad: {client['edad']}")
        print(f"Grupo edad: {client['grupo_edad']}")
        print(f"Código comportamiento: {client['codigo_comportamiento']}")
        print(f"Segmento: {client['segment']}")
        print(f"Mensaje: {client['missatge']}")
        print(f"Factor código: {client['factor_codi']}")
        print(f"Factor edad: {client['factor_edat']}")
        print(f"Riesgo final: {client['risc_final']:.3f}")
        print("-----------------------------------")
```

### Función `mostrar_respostes()`

1. Recibo la lista de clientes procesados con sus riesgos calculados.
2. Ordeno los clientes mediante `sorted()` y `reverse=True`, utilizando el riesgo final como criterio.
3. Muestro los tres clientes con mayor riesgo para identificar a quién contactar primero.
4. Selecciono los clientes que tienen el código de comportamiento 4 y comparo sus riesgos según el grupo de edad.
5. Identifico los clientes con códigos de comportamiento 0 y 1, ya que muestran menor avance en el proceso de compra.
6. Muestro al cliente con mayor riesgo como principal candidato para priorizar, aclarando que no disponemos de datos suficientes para determinar la rentabilidad económica real.

```python
def mostrar_respostes(resultats):

    ordenados = sorted(
        resultats,
        key=lambda client: client["risc_final"],
        reverse=True
    )

    # Pregunta 1
    print("\n1. CLIENTES A CONTACTAR PRIMERO")

    for client in ordenados[:3]:
        print(client["nombre"], "-", round(client["risc_final"], 3))

    # Pregunta 2
    print("\n2. MISMO COMPORTAMIENTO, DIFERENTE EDAD")

    for client in ordenados:
        if client["codigo_comportamiento"] == 4:
            print(
                client["nombre"],
                "-", client["grupo_edad"],
                "-", round(client["risc_final"], 3)
            )

    # Pregunta 3
    print("\n3. CLIENTES CON MENOR INTERÉS")

    for client in ordenados:
        if client["codigo_comportamiento"] in (0, 1):
            print(
                client["nombre"],
                "- Código:", client["codigo_comportamiento"],
                "- Riesgo:", round(client["risc_final"], 3)
            )

    # Pregunta 4
    print("\n4. CLIENTE CON MAYOR RIESGO")

    print(
        ordenados[0]["nombre"],
        "-", round(ordenados[0]["risc_final"], 3)
    )

    print("No podemos determinar la rentabilidad real sin datos económicos.")
```

```python
mostrar_respostes(resultats)
```

### Función `main_marketing()`

1. Recibo la lista original de clientes.
2. Llamo a `processar_clients()` para obtener la información de los clientes con sus segmentos, mensajes, factores y riesgos calculados.
3. Guardo los datos procesados en la variable `resultats`.
4. Llamo a `mostrar_diccionari()` para mostrar los resultados de los diez clientes.
5. Llamo a `mostrar_respostes()` para responder las cuatro preguntas del departamento de Marketing.

```python
def main_marketing(clients):

    resultats = processar_clients(clients)

    mostrar_diccionari(resultats)

    mostrar_respostes(resultats)
```
