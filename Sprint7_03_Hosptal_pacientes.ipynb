### Datos iniciales: Pacientes

Utilizo la lista de cinco pacientes proporcionada en el enunciado. Cada registro contiene el nombre, la edad y los síntomas en una misma cadena de texto, separados por el símbolo `|`.

Estos datos serán procesados posteriormente para separar la información, normalizar los síntomas y calcular el riesgo total de cada paciente.

```python
pacientes = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]
```

### Datos iniciales: Niveles de riesgo

Utilizo el diccionario de puntuaciones proporcionado en el enunciado, que asigna un valor numérico a cada síntoma.

Estos valores serán utilizados por la función `calcular_riesgo()` para sumar las puntuaciones correspondientes a los síntomas de cada paciente y obtener su riesgo total.

```python
niveles_riesgo = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

### Función `separar_datos()`

1. Recibo un registro de paciente en formato de texto.
2. Utilizo `split("|")` para separar el nombre, la edad y los síntomas.
3. Guardo los tres valores en variables independientes.
4. Convierto la edad a un número entero mediante `int()`.
5. Devuelvo el nombre, la edad y el texto de los síntomas para que puedan utilizarse en las siguientes funciones.

```python
def separar_datos(registro):

    nombre, edad, sintomas = registro.split("|")

    edad = int(edad)

    return nombre, edad, sintomas
```

Comprobación

```python
nombre, edad, sintomas = separar_datos(
    "Maria|45|tos, Fiebre Alta, dolor_pecho"
)

print("Nombre:", nombre)
print("Edad:", edad)
print("Síntomas:", sintomas)
```

### Función `normalizar_sintomas()`

1. Recibo el texto que contiene los síntomas del paciente.
2. Separo los síntomas por comas utilizando `split(",")`.
3. Creo una lista vacía para guardar los síntomas normalizados.
4. Recorro cada síntoma mediante un bucle `for`.
5. Elimino los espacios al principio y al final utilizando `strip()`.
6. Convierto cada síntoma a minúsculas mediante `lower()`.
7. Reemplazo los espacios interiores por guiones bajos utilizando `replace()`.
8. Agrego cada síntoma normalizado a la nueva lista mediante `append()`.
9. Devuelvo la lista completa de síntomas normalizados.

```python
def normalizar_sintomas(sintomas):

    lista_sintomas = sintomas.split(",")

    sintomas_normalizados = []

    for sintoma in lista_sintomas:

        sintoma = sintoma.strip().lower().replace(" ", "_")

        sintomas_normalizados.append(sintoma)

    return sintomas_normalizados
```

Comprobación

```python
resultado = normalizar_sintomas(
    "fiebre alta, dificultad RESPIRAR, tos"
)

print(resultado)
```

### Función `crear_diccionario()`

1. Recibo el nombre, la edad y la lista de síntomas normalizados de un paciente.
2. Creo un diccionario llamado `paciente`.
3. Guardo el nombre en la clave `"nombre"`.
4. Guardo la edad en la clave `"edad"`.
5. Guardo la lista de síntomas en la clave `"sintomas"`.
6. Devuelvo el diccionario completo mediante `return` para utilizarlo en las siguientes funciones.

```python
def crear_diccionario(nombre, edad, sintomas):

    paciente = {
        "nombre": nombre,
        "edad": edad,
        "sintomas": sintomas
    }

    return paciente
```

Comprobación



```python
paciente = crear_diccionario(
    "Maria",
    45,
    ["tos", "fiebre_alta", "dolor_pecho"]
)

print(paciente)
```

### Función `calcular_riesgo()`

1. Recibo la lista de síntomas normalizados y el diccionario con sus puntuaciones.
2. Creo una variable llamada `riesgo_total` y la inicializo en cero.
3. Recorro la lista de síntomas mediante un bucle `for`.
4. Para cada síntoma, busco su puntuación en el diccionario de riesgos.
5. Sumo la puntuación obtenida a la variable `riesgo_total`.
6. Cuando termino de recorrer todos los síntomas, devuelvo el riesgo total mediante `return`.

```python
def calcular_riesgo(sintomas, niveles_riesgo):

    riesgo_total = 0

    for sintoma in sintomas:
        riesgo_total += niveles_riesgo[sintoma]

    return riesgo_total
```

Comprobación

```python
riesgo = calcular_riesgo(
    ["tos", "fiebre_alta", "dolor_pecho"],
    niveles_riesgo
)

print("Riesgo total de Maria:", riesgo)
```

### Función `ordenar_pacientes()`

1. Recibo una lista de pacientes con sus riesgos totales calculados.
2. Utilizo `sorted()` para ordenar los pacientes según su riesgo total.
3. Indico mediante `key` que el criterio de ordenación será la clave `"riesgo_total"` de cada diccionario.
4. Utilizo `reverse=True` para ordenar las puntuaciones de mayor a menor.
5. Guardo el resultado en una nueva lista llamada `pacientes_ordenados`.
6. Devuelvo la lista ordenada mediante `return`.

```python
def ordenar_pacientes(pacientes):

    pacientes_ordenados = sorted(
        pacientes,
        key=lambda paciente: paciente["riesgo_total"],
        reverse=True
    )

    return pacientes_ordenados
```

Comprobación

```python
pacientes_prueba = [
    {"nombre": "Maria", "riesgo_total": 8},
    {"nombre": "Luis", "riesgo_total": 2},
    {"nombre": "Sara", "riesgo_total": 9}
]

resultado = ordenar_pacientes(pacientes_prueba)

for paciente in resultado:
    print(paciente["nombre"], paciente["riesgo_total"])
```

### Función `mostrar_resultados()`

1. Recibo la lista de pacientes ordenados según su riesgo total.
2. Muestro un encabezado para identificar el informe final.
3. Recorro la lista de pacientes utilizando un bucle `for`.
4. En cada iteración, obtengo el nombre, la edad, los síntomas normalizados y el riesgo total del paciente.
5. Muestro la información mediante `print()` y f-strings.
6. Agrego una separación entre pacientes para facilitar la lectura del informe.

```python
def mostrar_resultados(pacientes_ordenados):

    print("\n========== PRIORIDAD DE PACIENTES ==========")

    for paciente in pacientes_ordenados:

        print(f"Nombre: {paciente['nombre']}")
        print(f"Edad: {paciente['edad']}")
        print(f"Síntomas: {paciente['sintomas']}")
        print(f"Riesgo total: {paciente['riesgo_total']}")
        print("-------------------------------------------")
```

Comprobación

```python
pacientes_prueba = [
    {
        "nombre": "Sara",
        "edad": 67,
        "sintomas": ["fiebre_alta", "dificultad_respirar", "tos"],
        "riesgo_total": 9
    },
    {
        "nombre": "Maria",
        "edad": 45,
        "sintomas": ["tos", "fiebre_alta", "dolor_pecho"],
        "riesgo_total": 8
    }
]

mostrar_resultados(pacientes_prueba)
```

### Función `gestionar_pacientes()`

1. Recibo la lista original de pacientes y el diccionario con los niveles de riesgo.
2. Creo una lista vacía llamada `pacientes_procesados` para guardar los resultados.
3. Recorro cada registro de la lista de pacientes mediante un bucle `for`.
4. Utilizo `separar_datos()` para obtener el nombre, la edad y el texto de síntomas.
5. Utilizo `normalizar_sintomas()` para limpiar los síntomas y convertirlos en una lista.
6. Utilizo `crear_diccionario()` para organizar los datos de cada paciente.
7. Utilizo `calcular_riesgo()` para obtener la puntuación total de sus síntomas.
8. Guardo el riesgo calculado dentro del diccionario del paciente.
9. Agrego el diccionario completo a la lista `pacientes_procesados`.
10. Una vez procesados todos los pacientes, utilizo `ordenar_pacientes()` para organizarlos de mayor a menor riesgo.
11. Utilizo `mostrar_resultados()` para presentar el informe final.
12. Devuelvo la lista ordenada para poder reutilizarla si fuera necesario.

```python
def gestionar_pacientes(pacientes, niveles_riesgo):

    pacientes_procesados = []

    for registro in pacientes:

        nombre, edad, sintomas = separar_datos(registro)

        sintomas_limpios = normalizar_sintomas(sintomas)

        paciente = crear_diccionario(nombre, edad, sintomas_limpios)

        riesgo = calcular_riesgo(sintomas_limpios, niveles_riesgo)

        paciente["riesgo_total"] = riesgo

        pacientes_procesados.append(paciente)

    pacientes_ordenados = ordenar_pacientes(pacientes_procesados)

    mostrar_resultados(pacientes_ordenados)

    return pacientes_ordenados
```
